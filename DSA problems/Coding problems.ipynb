```python
nums = [2, 7, 11, 15]
target = 9
def two_sum(nums, target):
    seen = {}
    for i, num in enumerate(nums):
        diff = target - num
        if diff in seen:
            return [seen[diff], i]
        seen[num] = i
two_sum(nums,target)
# Time complexity ==>  O(n)
# Space complexity ==> O(n)
```

```python
arr = [2, 3, -2, 4]
cur_max = arr[0]
cur_min = arr[0]
result = arr[0]

for i in arr[1:]:
    temp_max = max(i, i*cur_max, i*cur_min)
    temp_min = min(i, i*cur_max, i*cur_min)

    cur_max = temp_max
    cur_min = temp_min
    result = max(cur_max, result)



print(result)
```

#### Merge Sort Algorithm o(nlogn) in all case

```python
def merge_sort(arr):
    # Base case
    if len(arr) <= 1:
        return arr

    # Divide
    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])

    # Merge
    return merge(left, right)


def merge(left, right):
    result = []
    i = j = 0

    # Compare elements and merge
    while i < len(left) and j < len(right):
        if left[i] < right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    # Add remaining elements
    result.extend(left[i:])
    result.extend(right[j:])

    return result


# Example usage
arr = [38, 27, 43, 3, 9, 82, 10]

sorted_arr = merge_sort(arr)

print("Original Array:", arr)
print("Sorted Array:", sorted_arr)
```

```python
def replaceSubstr(s, s1, s2):
    ans = ""
    n, m = len(s), len(s1)

    i = 0
    while i < n:
        if i + m <= n and s[i:i + m] == s1:
            ans += s2

            # Skip the characters of s1
            i += m - 1
        else:
            ans += s[i]
        i += 1
    return ans


s = "abababa"
s1 = "aba"
s2 = "a"
print(replaceSubstr(s, s1, s2))
```

BUY AND SELL STOCK ON MAXIMUM PROFIT 

You are given an array prices where prices[i] is the price of a given stock on the ith day.

You want to maximize your profit by choosing a single day to buy one stock and choosing a different day in the future to sell that stock.

Return the maximum profit you can achieve from this transaction. If you cannot achieve any profit, return 0.

 

Example 1:

Input: prices = [7,1,5,3,6,4]
Output: 5
Explanation: Buy on day 2 (price = 1) and sell on day 5 (price = 6), profit = 6-1 = 5.
Note that buying on day 2 and selling on day 1 is not allowed because you must buy before you sell.
Example 2:

Input: prices = [7,6,4,3,1]
Output: 0
Explanation: In this case, no transactions are done and the max profit = 0.

```python
prices = [7,5,1,3,6,4] 

n = len(prices) 
max_profit = 0
day = [0,0]
for starting_price_i in range(n):
    for closing_price_i in range(starting_price_i + 1,n):
        profit = prices[closing_price_i]  - prices[starting_price_i]
        if profit >=0 and profit > max_profit:
            max_profit = max(max_profit,profit)
            day[0] = starting_price_i
            day[1] = closing_price_i


print(f"buy on day  : {day[0]}    sell   ==>    {day[1]}")
print(max_profit)
```

```python
def maxProfit(prices):
    min_price = float('inf')   # sabse chhoti price track karega
    max_profit = 0             # maximum profit

    for price in prices:

        # agar current price aur chhoti hai
        if price < min_price:
            min_price = price

        # current day par sell karne ka profit
        profit = price - min_price

        # max profit update
        if profit > max_profit:
            max_profit = profit

    return max_profit


# Example 1
prices = [7,1,5,3,6,4]
print(maxProfit(prices))   # 5

# Example 2
prices = [7,6,4,3,1]
print(maxProfit(prices))   # 0
```

       FIND LONGEST COMMON PREFIX           
Write a function to find the longest common prefix string amongst an array of strings.

If there is no common prefix, return an empty string "".

Example 1:

Input: strs = ["flower","flow","flight"]
Output: "fl"
Example 2:

Input: strs = ["dog","racecar","car"]
Output: ""
Explanation: There is no common prefix among the input strings.

```python
strs = ["flower","flow","flight"] 
def longest_common_prefix(strs):
    min_len_word = min(list(map(len,strs)))

    result = []

    for i in range(min_len_word):
        char = strs[0][i]

        for word in strs:
            if word[i] != char:
                return "".join(result)
        result.append(char)
    return "".join(result)


longest_common_prefix(strs)

```

## -------- FLATTEN ARRAY PROBLEM -----------


```python
def flatten(lst):
    result = []

    for item in lst:
        if isinstance(item, list):
            result.extend(flatten(item))
        else:
            result.append(item)

    return result


nested = [1, [2, 3], [4, [5, 6], 7], 8]
print(flatten(nested))
```
