###### SOLVING USING 2 POINTER (SLOW & FAST)

```python
arr = [0,1,0,3,0,12] 
slow = 0 
for fast in range(len(arr)):
    if arr[fast] != 0:
        arr[slow],arr[fast] = arr[fast] , arr[slow]
        slow += 1  
print(arr)    
```

```python
# arr = [1,[2,[3,5]],6]  <-- INPUT 
# arr = [1,2,3,5,6]      <-- OUTPUT
# USING RECURSION  

arr = [1,[2,[3,5]],6]
def flatten(arr):
    result = []
    for item in arr :
        if isinstance(item,list):
            result.extend(flatten(item)) 
        else:
            result.append(item) 
        # print(result)
    return result 
print("INPUT  : ",arr)
print("OUTPUT :",flatten(arr))

```

```python
# arr = [1,[2,[3,5]],6]  <-- INPUT 
# arr = [1,2,3,5,6]      <-- OUTPUT
# USING STACK  

arr = [1,[2,[3,5]],6]
def flatten_with_stack(arr):
    stack = list(arr)
    result = []
    while stack:
        val = stack.pop()
        if isinstance(val,list):
            stack.extend(val)
        else:
            result.append(val) 
    return result[::-1]

print("INPUT  : ",arr)
print("OUTPUT :",flatten_with_stack(arr))
```
